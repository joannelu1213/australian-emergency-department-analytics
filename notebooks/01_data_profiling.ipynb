# Australian Emergency Department Data Profiling


## Purpose

This notebook examines the structure, coverage, data types, reporting levels and data-quality issues in the AIHW emergency department extract before cleaning and modelling.

```python
from pathlib import Path

import pandas as pd
```

```python
file_path = Path("../data/raw/myhosp-emergency-department-data-extract.xlsx")

if not file_path.exists():
    raise FileNotFoundError(f"Data not found: {file_path.resolve()}")

file_path
```

```python
data = pd.ExcelFile(file_path)

data.sheet_names
```

```python
# Inspect the first 15 row of each data sheet
for sheet_name in data.sheet_names:
    preview = pd.read_excel(
        file_path,
        sheet_name=sheet_name,
        header=None,
        nrows=30,
    )

    print(f"{sheet_name}: {preview.shape[0]} preview rows, {preview.shape[1]} columns")
    display(preview)
```

```python
# Skip the non-structured data rows, each sheet contains only the column headers and the main data entry
sheet_config = {
    "Patients seen on time": 16,
    "Time in ED - within 4 hrs": 19,
    "Time in ED": 16,
    "Presentations": 15,
}
```

```python
processed_data_v1 = {}

for sheet_name, rows_to_skip in sheet_config.items():
    processed_data_v1[sheet_name] = pd.read_excel(
        file_path,
        sheet_name=sheet_name,
        skiprows=rows_to_skip
    )

for sheet_name, df in processed_data_v1.items():
    print(f"\n--- {sheet_name} ---")
    print(df.shape)
    display(df.head())
```

```python
# Data Inspection

for sheet_name, df in processed_data_v1.items():
    print(f"\n{'=' * 60}")
    print(f"Sheet: {sheet_name}")
    print(f"{'=' * 60}")

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nMissing values:")
    print(df.isna().sum())

    print("\nUnique values per column:")
    print(df.nunique(dropna=False))

    print("\nFirst 5 rows:")
    display(df.head())
```

```python
# Assign names for each data sheet

presentations = processed_data_v1["Presentations"]
seen_on_time = processed_data_v1["Patients seen on time"]
within_4hrs = processed_data_v1["Time in ED - within 4 hrs"]
time_in_ed = processed_data_v1["Time in ED"]

```

```python
## Check data quality 

# 1. presentations 

# categorical inspection
for col in [
    "Reporting unit type",
    "State",
    "Year",
    "Triage category"
]:
    print(f"\n--- {col} ---")
    print(presentations[col].value_counts(dropna=False))
```

```python
# inspect the measure column
presentations["Number of presentations"].value_counts(
    dropna=False
).head(30)
```

```python
non_numeric_values = presentations.loc[
    presentations["Number of presentations"].notna()
    & pd.to_numeric(
        presentations["Number of presentations"],
        errors="coerce"
    ).isna(),
    "Number of presentations"
]

non_numeric_values.value_counts(dropna=False)
```

```python
# Missing values in the measure column
missing_presentations = presentations[
    presentations["Number of presentations"].isna()
]

print("Missing rows:", len(missing_presentations))
```

Number of presentations contains no missing values. However, 616 records are reported as <5, which prevents the column from being stored as numeric. These small-count values will require explicit treatment during cleaning.

```python
# Check if there is any duplications
candidate_key = [
    "Reporting unit",
    "Reporting unit type",
    "State",
    "Year",
    "Triage category"
]

presentations.duplicated(
    subset=candidate_key,
    keep=False
).sum()

```

```python
duration_pattern = r"^\d+\s+hrs?\s+\d+\s+mins?$"

for col in [
    "Median time",
    "Time until most patients (90%) depart ED",
    "Peer group average (90%)"
]:
    special_values = time_in_ed.loc[
        time_in_ed[col].notna()
        & ~time_in_ed[col].astype(str).str.match(duration_pattern),
        col
    ]

    print(f"\n--- Special values in {col} ---")
    print(special_values.value_counts(dropna=False))
```

Main Findings:
- Unnamed: 7 is completely empty -> safe to drop later.
- Unnamed: 9 contains 94 * flags.
- Unnamed: 11 also contains 94 * flags.
- Median time and Time until most patients (90%) depart ED use consistent duration strings such as 2 hrs 58 mins.
- Both duration fields also contain the publication codes –, NP, and NP†.
- Peer group average (90%) contains Not peered for 1,716 rows.
- Peer-group averages are structurally missing for National and Local Hospital Network rows.
- Not peered occurs only for hospital records.
- Candidate-key duplicates = 0.
- Exact duplicates = 0.
